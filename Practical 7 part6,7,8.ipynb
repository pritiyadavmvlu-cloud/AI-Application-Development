```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
# Load dataset
data = pd.read_csv("iris.csv")
# Input and output
X = data[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = data['species']
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1
)
# Create model
model = DecisionTreeClassifier()
# Train model
model.fit(X_train, y_train)
# Display accuracy
print("Training Accuracy:", model.score(X_train, y_train) * 100)
print("Testing Accuracy:", model.score(X_test, y_test) * 100)
# Prediction
new_data = pd.DataFrame([[5.1, 3.5, 1.4, 0.2]],
                        columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
prediction = model.predict(new_data)
print("Predicted Species:", prediction[0])
print("Priti Yadav T054")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
# Load dataset
data = pd.read_csv("NetFlix.csv")
# Display columns (important to verify)
print(data.columns)
# Clean column names
data.columns = data.columns.str.strip()
# Encode categorical columns
le = LabelEncoder()
# Example: Using 'type' as target (Movie/TV Show)
data['type'] = le.fit_transform(data['type'])
# Select features (modify based on your dataset)
# Here we use some common columns
features = ['release_year']  # numeric column
X = data[features]
y = data['type']
# Handle missing values
X.fillna(X.mean(numeric_only=True), inplace=True)
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1)
# Create model
model = DecisionTreeClassifier()
# Train model
model.fit(X_train, y_train)
# Display accuracy
print("Training Accuracy:", model.score(X_train, y_train) * 100)
print("Testing Accuracy:", model.score(X_test, y_test) * 100)
# Prediction (example: release year = 2020)
new_data = pd.DataFrame([[2020]], columns=['release_year'])
prediction = model.predict(new_data)
# Decode prediction
if prediction[0] == 0:
    print("Predicted Type: Movie")
else:
    print("Predicted Type: TV Show")
print("Priti Yadav T054")
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
# Load dataset
iris = load_iris()
X = iris.data
y = iris.target
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1
)
# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)
# Train model
model.fit(X_train, y_train)
# Display accuracy
print("Training Accuracy:", model.score(X_train, y_train)*100)
print("Testing Accuracy:", model.score(X_test, y_test)*100)
# Prediction
prediction = model.predict([X_test[0]])
print("Predicted Class:", prediction[0])
print("Priti Yadav T054")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder
# Load dataset
data = pd.read_csv("NetFlix.csv")
# Clean column names
data.columns = data.columns.str.strip()
# Encode categorical columns
le = LabelEncoder()
# Target column: type (Movie / TV Show)
data['type'] = le.fit_transform(data['type'])
# Feature selection (numeric column)
X = data[['release_year']]
y = data['type']
# Handle missing values
X = X.fillna(X.mean(numeric_only=True))
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1
)
# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)
# Train model
model.fit(X_train, y_train)
# Display accuracy
print("Training Accuracy:", model.score(X_train, y_train) * 100)
print("Testing Accuracy:", model.score(X_test, y_test) * 100)
# Prediction (example input)
prediction = model.predict([[2020]])
# Decode output
if prediction[0] == 0:
    print("Predicted Type: Movie")
else:
    print("Predicted Type: TV Show")
print("Priti Yadav T054")
```
